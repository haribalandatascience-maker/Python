What is Python?

Python is a high-level, interpreted, general -purpose programming language.
It was created by Guido van Rossum and first released in 1991.
Python is known for its simplicity, readability and productivity.

★ How Python Works?

1. Write code in Python file (.py)
2. Python Interpreter reads the code
3. Interpreter converts it to bytecode
4. Python Virtual Machine (PVM) executes it
5. Output is displayed on screen 

What is a Variable ?

A variable is a named location used to store data in memory.
In Python, you do not need to declare the type of variable explicitly.
